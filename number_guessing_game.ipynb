```python
import random

number = random.randint(1,10)
tries = 3

for attempt in range(tries):
  guess = int(input(f"guess a number(1-10) - {tries - attempt} tries left:"))
  if guess == number:
    print("you won🏆!")
    break
  elif guess < number:
    print("too low! 🔽")
  else:
    print("too high!🔼")
else:
  print("game over😭! the number was",number)
```
